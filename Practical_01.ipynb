A] Simple Linear Regression by Using Scikit-learn

```python
x = df[['Study_Hours']] # Independent Variable
y = df['Marks'] # Dependent Variable
```

```python
x
```

4] Model Building

```python
#Split dataset into training and testing set
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)
```

```python
x_test
```

```python
#Import Model
from sklearn.linear_model import LinearRegression
model = LinearRegression()

#Train the model
model.fit(x_train, y_train)
```

```python
#Predict the values
y_pred = model.predict(x_test)
y_pred
```

```python
#Print Equation
print('Intercept:', model.intercept_)
print('Slope:', model.coef_[0])
```

SLR Equation

* Marks = (6.48*Study_Hours) + 28.27



5] Model Evaluation

```python
# Evaluation
from sklearn.metrics import root_mean_squared_error, r2_score, mean_squared_error
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print('RMSE:', round(rmse, 3))
print('R2 Score:', round(r2, 2))
```

```python
#Regression Line
plt.figure(figsize=(7, 5))
plt.scatter(x, y, color="blue", label='Actual Data')
plt.plot( x_test, y_pred,
         color='red', label='Regression Line')

plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Simple Linear Regression')
plt.legend()
plt.show()
```

Residual Plot

```python
residuals = y_test - y_pred # Calculate Residual
plt.figure(figsize=(7, 5))
plt.scatter( y_pred, residuals, color='green' )
plt.axhline( y=0, color="red", linestyle="--" )
plt.xlabel('Predicted Marks')
plt.ylabel('Residuals')
plt.title('Residual Plot')
plt.show()
```

B] Simple Linear Regression by Using Numpy

```python
x = df["Study_Hours"].values # Independent Variable (Features)
y = df["Marks"].values # Dependent Variable (Target)
```

```python
x
```

```python
#Calculate Mean
x_mean = np.mean(x)
y_mean = np.mean(y)
```

```python
x_mean
```

```python
# Calculate Slope
m = np.sum((x-x_mean)*(y-y_mean))/np.sum((x-x_mean)**2)

# Calculate Intercept
b = y_mean - m*x_mean

print("Slope =", m)
print("Intercept =", b)
```

```python
#Prediction
y_pred_numpy = m * x + b
y_pred_numpy
```

C] Multiple Linear Regression

```python
#Feature Selection
x = df[['Study_Hours', 'Attendance']]  # Independent Variables
y = df['Marks']                         # Dependent Variable

#Split Dataset
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3,
                                                    random_state=42)
```

```python
x_train
```

Model Building

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
#Prediction
y_pred = model.predict(x_test)
y_pred
```

```python
#Model Parameters
print('Intercept:', model.intercept_)
print("Coefficients")
coeff = pd.DataFrame({
    'Feature': x.columns, 'Coefficient': model.coef_ })

print(coeff)
```

Model Evaluation

```python
# Evaluation Metrics
from sklearn.metrics import mean_absolute_error

print('Metrics Values of MLR:')
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

print('MAE:', mae)
print('RMSE:', round(rmse, 2))
print('R2 Score:', round(r2, 2))
```

```python
# Actual vs Predicted Plot
plt.figure(figsize=(7, 5))
plt.scatter(y_test, y_pred, color='Purple')

# Perfect Prediction Line
plt.plot([y.min(), y.max()],
         [y_pred.min(), y_pred.max()],color='red')

plt.xlabel('Actual Marks')
plt.ylabel('Predicted Marks')
plt.title('Actual vs Predicted (MLR)')
plt.show()
```

Residual Plot

```python
residuals = y_test - y_pred # calculate residual

plt.scatter( y_pred, residuals, color="Orange" )
plt.axhline( y=0, color='red', linestyle='--' )
plt.xlabel('Predicted Marks')
plt.ylabel('Residuals')
plt.title('Residual Plot (MLR)')
plt.show()
```
